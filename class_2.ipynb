## Mini Challenge 2: Given a list of number determine which on is the largest

Given a collection of numbers, implement an algorithm that will help us to determing which is the largest number. 

Criteria:

-Create a function that recieves a collection and returns a number
-Keep edges cases in mind: if the collection is emputy I will return a string that says "No values to compare"

The collections is numbers = [12,4,27,3,86,34,15]

```python
numbers = [12, 4, 27, 3, 86, 34, 15]

def find_largest(collection):
    if len(collection) == 0:
        return "No values to compare"

    largest = collection[0]

    for number in collection:
        if number > largest:
            largest = number

    return largest


print(find_largest(numbers))
```

```python
## mini Challenge 3: Anagram Check

Create a function that takes two strings as parameters and returns true if they are anagrams or each other, False otherwise.

Examples:
1. heart and earth
2. scar and cars
3. star and rats
etc


## Criteria
Two stings are anagrams of each other if and only if all of the criteria below are met:

1. They must use all the same characters
2.They must be of exactly the same lenght (in characters)
3.Sorted function is not allowed
```

```python
def are_anagrams(word1, word2):

    
    if len(word1) != len(word2):
        return False

    
    count1 = {}
    count2 = {}

    # Count characters in word1
    for letter in word1:
        if letter in count1:
            count1[letter] += 1
        else:
            count1[letter] = 1

    # Count characters in word2
    for letter in word2:
        if letter in count2:
            count2[letter] += 1
        else:
            count2[letter] = 1

    # Compare the dictionaries
    if count1 == count2:
        return True
    else:
        return False


print(are_anagrams("heart", "earth"))
print(are_anagrams("scar", "cars"))
print(are_anagrams("star", "rats"))
print(are_anagrams("hello", "world"))
```

```python
class Calculator:

    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2

    def subtract(self):
        return self.num1 - self.num2

    def multiply(self):
        return self.num1 * self.num2

    def divide(self):
        if self.num2 == 0:
            return "Cannot divide by zero"
        
        return self.num1 / self.num2

    def __str__(self):
        return f"Calculator({self.num1}, {self.num2})"


c1 = Calculator(5, 2)

print(c1)
print(c1.add())
print(c1.subtract())
print(c1.multiply())
print(c1.divide())
```
