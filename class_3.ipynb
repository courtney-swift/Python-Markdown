## Stacks
A stack is an ordered collection of elements where items are added and removed from one end, commonly referred to as the "top". The ordering principle is referred to as LIFO (Last In, First Out)

```python
# Simplified Version (relying on built-ins fuctions)

class stack:
  def __init__(self):
    self.items = []

  def push(self, elements):
    self.items.append(elements)

  def pop(self):
    return self.items.pop()

  def peek(self):
    return self.items[len(self.items)-1]

  def is_empty(self):
      return self.items ==[]

  def __str__(self):
    return f"Stack - {self.items}"
```

```python
s1 = stack()
s1.push("A")
s1.push("B")
s1.push("C")
print(s1)
print(s1.peek())
print(s1)
hold = s1.pop()
print(hold)
print(s1)
s1.pop()
s1.pop()
print(s1)
print(s1.is_empty())
```
