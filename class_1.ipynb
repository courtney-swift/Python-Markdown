```python
# OOP (Object Oriented Programming)
# Python is an object-oriented prrogramming, allowing us to structure code using classes and objects for a better organization and reuseability.

  
"""
Advantages of OOP:
- Provides clear structure for programs
- Makes code easier to maintain, reuse and debug
- Helps keep your code DRY (Do Not Repeat Yourself)
- Allows you to build reusable applications with less code

Classes and Objects:
A class defines what an object should look like and its behavior.
An object is created based on that class.

For example:
- class: Fruits -> objects: Apple, Banana, Mango
- class: Car -> objects: Audi, Tesla, Toyota

The Four Pillars of OOP:
1. Encapsulation
2. Abstraction
3. Inheritance
4. Polymorphism
"""
  
```

```python
class Person:
  # Construtor
  def __init__(self, name, last_name, age):
    self.name = name
    self.last_name = last_name
    self.age = age

  def greeting(self):
    return f"Hello from {self.name}"

  def __str__(self):
    return f"Person (name = {self.name}, last_name = {self.last_name}, age = {self.age})"
```

```python
p1 = Person("Courtney", "Swift", 36)
print(p1)
print(p1.name)
print(p1.greeting())
```
